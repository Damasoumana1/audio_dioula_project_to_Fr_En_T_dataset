# Analyse Audio
Écoute et visualisation de l'onde sonore d'un extrait Dioula.

```python
import librosa
import librosa.display
import matplotlib.pyplot as plt
import IPython.display as ipd
import os
import glob
```

```python
# Trouver un fichier audio
audio_files = glob.glob('../data/audio/wav/*.wav')
if audio_files:
    sample_audio = audio_files[0]
    print(f'Fichier analysé : {sample_audio}')
else:
    sample_audio = None
    print('Aucun fichier audio trouvé. Exécutez main.py.')
```

```python
if sample_audio:
    # Lecture de l'audio
    y, sr = librosa.load(sample_audio, sr=16000)
    
    # Affichage de l'onde
    plt.figure(figsize=(14, 5))
    librosa.display.waveshow(y, sr=sr)
    plt.title('Forme d\'onde (Waveform)')
    plt.xlabel('Temps (s)')
    plt.ylabel('Amplitude')
    plt.show()
    
    # Lecteur audio
    display(ipd.Audio(sample_audio))
```
